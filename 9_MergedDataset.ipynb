```python
# Install & Import
import pandas as pd
from google.colab import files

# Step 1: Upload Two CSV Files
print("Upload students.csv and grades.csv")
uploaded = files.upload()  # Upload both CSV files when prompted
```

```python
# Step 2: Read the uploaded files
students = pd.read_csv('students.csv')
grades = pd.read_csv('grades.csv')

# Preview
print("Students Data:")
print(students.head())

print("\nGrades Data:")
print(grades.head())
```

This joins/merges the two tables using the column student_id.

how='inner' means:

Only include rows where a student_id is present in both files.

You can change this to:

'left': All students, even if they don’t have grades

'outer': All rows from both files, missing values filled with NaN

```python
# Step 3: Merge the files
merged = pd.merge(students, grades, on='student_id', how='inner')  # change to 'left', 'outer' if needed

# Show the merged result
print("Merged Data:")
print(merged.head())
```

```python
# Step 4: Save and download the merged file
merged.to_csv('merged_data.csv', index=False)

# Download the file
files.download('merged_data.csv')
```
